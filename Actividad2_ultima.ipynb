# **2. LIMPIEZA DE DATOS**

```python
df["Animal type"] = (
    df["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["Animal type"].value_counts()
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirel", "Red Squirrell", regex=False)
    .str.replace("Red Squirrel", "Red Squirrell", regex=False)
    .str.replace("Lynx?", "Lynx", regex=False)
    .str.replace("European Bison™", "European Bisson", regex=False)
    .str.replace("European Bison", "European Bisson", regex=False)
    .str.replace("European Buster", "European Bisson", regex=False)
    .str.strip()
)

df["Animal type"].value_counts()
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirrelll", "Red Squirrell", regex=False)
    .str.replace("Red Squirrellll", "Red Squirrell", regex=False)
    .str.strip()
)

df["Animal type"].value_counts()
```

```python
 mode_Animal = df["Animal type"].mode()

df["Animal type"] = df["Animal type"].fillna(mode_Animal)

mode_Animal
```

```python
df["Animal type"].isnull().sum()
```

```python
df["Animal type"] = df["Animal type"].fillna("Desconocido")
```

```python
df["Animal type"].unique() # OK
```

```python
# Cuál es el promedio de una variable numérica según el tipo de animal?

promedio = df.groupby("Animal type")['Latitude'].mean()
print(promedio)

#El promedio de la altura de los tipos de animales es el: Bisonte Europeo
```

```python
# Cuál es el promedio de una variable numérica según el país?
promedio = df.groupby("Country")['Latitude'].mean()
print(promedio)

# De los paises que más tiene movimiento es: Polonia
```
